#### 1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
nums = [11,2,3,41,3] # creates a list of 5 integers
print("creates a list")
print(nums)
nums.append(10) # adds integer 10 at the end of the list
print("appends 10")
print(nums)
```

```python
nums.extend([12,14,16]) # extends or increases the number of elements in the list by adding integrs 12,14,16
print("add's 12,14,16")
print(nums)
```

```python
nums.remove(10) # finds and deletes the integer 10 from the list
print("removes 10")
print(nums)
```

```python
nums.pop() # deletes the last integer or element from tje list
print("removes last element")
print(nums)
```

```python
element = nums[4] # extracts integer in the place of 4th index and stoes it in variable 'element'
print("element from 4th index")
print(element)
```

```python
count_of_3 = nums.count(3) # counts how many times integer 3 is there in the list
print("3 is present this meny time in list")
print(count_of_3)
```

```python
nums.sort() # sorts the elements in ascending order
print("sorted in ascending order")
print(nums)
```

```python
nums.reverse() # reverse's the elements of list
print("reversed elements")
print(nums)
```

```python
nums.clear() # delets all the elements of list and keeps the list empty
print("empty list")
print(nums)
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
data = [
    {"Name": "Alice", "Age": 25, "City": "New York", "Salary": 70000},
    {"Name": "Alice", "Age": 25, "City": "New York", "Salary": 70000},
    {"Name": "Bob", "Age": 30, "City": "Chicago", "Salary": 85000},
    {"Name": "Charlie", "Age": 28, "City": "San Francisco", "Salary": 95000},
    {"Name": "Diana", "Age": 35, "City": "Austin", "Salary": 105000},
    {"Name": "Ethan", "Age": 40, "City": "Seattle", "Salary": 120000},
    {"Name": "Fiona", "Age": 22, "City": "Boston", "Salary": 65000},
    {"Name": "George", "Age": 31, "City": "Denver", "Salary": 78000},
    {"Name": "Hannah", "Age": 27, "City": "Miami", "Salary": 72000},
    {"Name": "Ian", "Age": 45, "City": "Los Angeles", "Salary": 135000},
    {"Name": "Julia", "Age": 29, "City": "Atlanta", "Salary": 80000},
]

df = pd.DataFrame(data)
print(f"{df}\n")

# created the dataframe with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'.
```

```python
print("")
df.info()
```

```python
# discribe gives the statistical measures of the data
df.describe()
```

```python
# prints two columns from the dataframe 
df[["Name", "City"]]
```

```python
# this will drop the column if it exists or do nothing if it dosent exists
df = df.drop("City", axis=1, errors="ignore")
df
```

```python
# adds a nan value
df.loc[2, "Salary"] = np.nan
df
```

```python
mean_salary = df["Salary"].mean() # calculates the mean value
df["Salary"] = df["Salary"].fillna(mean_salary) # and puts it in a nan value
df
```

```python
# delets the duplicate rows
df = df.drop_duplicates()
df
```
