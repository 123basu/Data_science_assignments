## SVM

- Support Vector Machine (SVM) is a supervised machine learning algorithm used for both classification and regression, though it is most commonly used for classification problems.

1. Intuition Behind SVM

Imagine you have two groups of data points:
- Red points = Cats
- Blue points = Dogs

The goal is to draw a line that separates the two groups.

Many lines can separate the classes, but SVM chooses the best possible line.

Instead of choosing any separating line, SVM chooses the line that leaves the maximum margin between the classes.

2. Margin

The margin is the distance between:
- The separating line (decision boundary)
- The nearest data points from each class

SVM tries to maximize this distance.

A larger margin generally means:
- Better generalization
- Less overfitting
- Better performance on unseen data

3. Hyperplane

In SVM, the separating boundary is called a hyperplane.

In 2D, A hyperplane is simply a line.

In 3D, A hyperplane becomes a plane.

In Higher Dimensions, It's called a hyperplane.


4. What Are Support Vectors?

The points closest to the hyperplane are called support vectors.

These points "support" the hyperplane, which is why the algorithm is called Support Vector Machine.

Important:
- SVM only depends heavily on these support vectors.
- Other points have little influence on the boundary.

7. Kernel Trick

When data can be separated by a straight line, we use Linear SVM.

What If Data Is Not Linearly Separable, No straight line can separate these classes. SVM solves this using the Kernel Trick.

The kernel trick transforms data into a higher-dimensional space where it becomes separable.
