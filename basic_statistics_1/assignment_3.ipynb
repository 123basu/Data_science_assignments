#### Bar Chart Analysis for Categorical Column

```python
df[['Day', 'SKU', 'City', 'BU', 'Brand', 'Model']]
```

```python
brand_counts = df['Brand'].value_counts()

plt.figure(figsize=(8,5))
plt.bar(brand_counts.index, brand_counts.values)
plt.title('Brand Frequency')
plt.xlabel('Brand')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

inferences
- The bar chart shows the distribution of products across different brands. Some brands have substantially more products or transactions than others.

```python
bu_counts = df['BU'].value_counts()

plt.figure(figsize=(6,4))
plt.bar(bu_counts.index, bu_counts.values)
plt.title('Business Unit Distribution')
plt.xlabel('BU')
plt.ylabel('Count')
plt.show()
```

inferences
- The dataset contains transactions from multiple business units, and they are kind of distributed equaly

```python
city_counts = df['City'].value_counts()

plt.figure(figsize=(6,4))
plt.bar(city_counts.index, city_counts.values)
plt.title('City Distribution')
plt.xlabel('City')
plt.ylabel('Count')
plt.show()
```

inferences
- all data belongs to one city 'C'.

```python
day_counts = df['Day'].value_counts()

plt.figure(figsize=(8,5))
plt.bar(day_counts.index, day_counts.values)
plt.title('Transactions by Day')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()
```

- Transaction frequencies vary across different days, indicating fluctuations in customer activity during the week. But only thursday has more customers compared to other days of the week

- Columns like SKU and Model may have too many unique values.

### Conclusion

- The descriptive analytics and data visualizations provide comprehensive insights into the sales dataset and reveal significant patterns in customer transactions, pricing, and sales performance.
- Analysis of the numerical variables indicates that columns such as Avg Price, Total Sales Value, Discount Amount, and Net Sales Value exhibit substantial variability. This is obsurved by large standard deviation values and differences between mean and median values. The distributions of these variables are positively skewed, suggesting that while the majority of transactions involve relatively lower values, a small number of transactions contribute exceptionally high sales and pricing figures.
- Histogram analysis confirmed the presence of right-skewed distributions, where most observations are concentrated within lower ranges and a few extreme values extend toward the higher end. Boxplot visualizations identified several upper-end outliers across major financial variables, indicating the existence of unusually large transactions and high-priced products within the dataset.
- Categorical analysis using bar charts highlighted variations in transaction frequencies across attributes such as Brand, Business Unit (BU), City, and Day. Certain categories appeared more dominant, indicating uneven distribution of sales activity among brands and business segments.
